# Тензоры

```python
import torch
import numpy as np
```

## Инициализация тензора

### Из данных

```python
data = [
    [1, 2],
    [3, 4]
]
tensor_1 = torch.tensor(data) # dtype = torch.int64
tensor_1
```

### Из массива numpy

```python
np_array = np.array(data)
tensor_2 = torch.tensor(np_array) # dtype = torch.int64
tensor_2
```

### Из другого тензора

```python
ones_tensor = torch.ones_like(tensor_2)
rand_tensor = torch.rand_like(tensor_1, dtype=torch.float32)
print(f'ones_tensor:\n{ones_tensor}')
print(f'rand_tensor:\n{rand_tensor=}')
```

### Случайные или постоянные значения

```python
shape = (2, 3)
rand_tensor = torch.rand(shape)
zeros_tensor = torch.zeros(shape)
ones_tensor = torch.ones(shape)

print(f'rand:\n{rand_tensor}')
print(f'zeros:\n{zeros_tensor}')
print(f'ones:\n{ones_tensor}')
```

## Атрибуты тензора

```python
tensor = torch.rand(3, 4)

print(f'shape: {tensor.shape}')
print(f'size: {tensor.size()}')
print(f'ndim: {tensor.ndim}')
print(f'type: {tensor.dtype}')
print(f'device: {tensor.device}')
```

## Операции на тензорах

### Индексация

```python
tensor = torch.ones(5, 5)

print(f'First row: {tensor[0]}')
print(f'First col: {tensor[:, 0]}')
print(f'Last col: {tensor[:, -1]}')
tensor[...,0] = 0
print(tensor)
```

### Конкатенация

```python
t = torch.cat([tensor, tensor], dim=1)
t
```

### Арифметические операции

```python
t1 = torch.rand((2, 3))
t2 = torch.rand((3, 4))
print(f't1:\n{t1}')
print(f't2:\n{t2}')

t3 = t1 @ t2
t4 = t1.matmul(t2)
print(f't1 @ t2 =\n{t3}')
print(f't1.matmul(t.2) = \n{t4}')

t5 = t1 * t1
t6 = t1.mul(t1)
print(f't1 * t1 =\n{t5}')
print(f't1.mul(t1) = \n{t6}')

```

### item()

```python
s = t1.sum()
s.item()
```

### Операции на месте

Операции, которые хранят результат в операнде, называются на месте. Они обозначаются через _ в конце. Например, к примеру: x.copy_(y), x.t_(), изменится x.

```python
tensor = torch.rand(2,2)
print(tensor)
tensor.add_(10)
print(f'After tensor.add_(10):\n{tensor}')
tensor.add(100)
print(f'After tensor.add(10):\n{tensor}')
```

## Связь с Numpy

### Тензор в Numpy массив

Изменение тензора отражается в массиве NumPy.

```python
t = torch.rand(2,2)
n = t.numpy()

print(t)
print(n)

t.add_(10)
print(t)
print(n)
```

### Массив NumPy для Tensor 

```python
n = np.ones(5)
t = torch.from_numpy(n)
print(t)
print(n)
```
