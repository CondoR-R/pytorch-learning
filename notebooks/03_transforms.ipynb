# Трансформации

Данные не всегда поступают в окончательную обработанную форму, которая требуется для обучения алгоритмов машинного обучения. Мы используем преобразования для выполнения некоторых манипуляций с данными и делаем их пригодными для обучения.

Все наборы данных TorchVision имеют два параметра - `transform` для изменения функций и `target_transform` для модификации меток - которые принимают вызываемые, содержащие логику преобразования. Модуль `Torchvision.transforms` предлагает несколько часто используемых трансформаций из коробки.

Функции FashionMNIST находятся в формате PIL Image, а метки являются целыми числами. Для обучения нам нужны функции как нормализованные тензоры, так и метки как one-hot вектора. Чтобы сделать эти преобразования, мы используем `torchvision.transforms.v2` API вместе с `torch.nn.functional.one_hot`.

```python
import torch
import torch.nn.functional as F
from torchvision import datasets
from torchvision.transforms import v2

ds = datasets.FashionMNIST(
    root="data",
    train=True,
    download=True,
    transform=v2.Compose([v2.ToImage(), v2.ToDtype(torch.float32, scale=True)]),
    target_transform=v2.Lambda(
        lambda y: F.one_hot(torch.tensor(y), num_classes=10).float()
    ),
)
```
